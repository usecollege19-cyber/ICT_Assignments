# 1.FizzBuzz

```python
n=int(input('Enter the Range '))
for n in range(1,n+1):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

# 2.Electricity Bill Calculator


```python
u=int(input('Enter the Units Consumed '))
if u<=100:
  bill=u*5
elif u<=200:
  bill=(100*5)+(u-100)*7
else:
  bill=(100*5)+(100*7)+(u-200)*10
print('Total Bill Amount: ',bill)
```

# 4.Palindrome

```python
a=input('Enter the String ')
b=a[::-1]
if a==b:
  print('The given number/string is Palindrome')
else:
  print('The given number/string is not Palindrome')
```
