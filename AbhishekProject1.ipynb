```python
inventory = {
    "Pen": {"price": 10, "stock": 20},
    "Notebook": {"price": 50, "stock": 10}
}

TAX = 0.05
LOW_STOCK = 3

# Add product (Admin)
def add_product():
    name = input("Product name: ")
    price = float(input("Price: "))
    stock = int(input("Stock: "))
    if stock < 0:
        print("Stock cannot be negative")
        return
    inventory[name] = {"price": price, "stock": stock}

# Purchase products (Customer)
def purchase():
    cart = []
    while True:
        name = input("Product (or done): ")
        if name == "done":
            break
        if name not in inventory:
            print("Not available")
            continue

        qty = int(input("Quantity: "))
        if qty > inventory[name]["stock"]:
            print("Not enough stock")
            continue

        cart.append((name, qty))

    bill(cart)

def bill(cart):
    total = 0
    print("\n--- BILL ---")
    for item, qty in cart:
        cost = inventory[item]["price"] * qty
        inventory[item]["stock"] -= qty
        total += cost
        print(item, qty, "=", cost)

        if inventory[item]["stock"] <= LOW_STOCK:
            print("Low stock alert for", item)

    discount = 0
    if total >= 1000:
        discount = total * 0.10
    elif total >= 500:
        discount = total * 0.05

    tax = (total - discount) * TAX
    final_amount = total - discount + tax

    print("Subtotal:", total)
    print("Discount:", discount)
    print("Tax:", tax)
    print("Payable:", final_amount)

# Main menu
while True:
    print("\n1.Add Product")
    print("2.Purchase")
    print("3.Exit")

    choice = input("Choice: ")

    if choice == "1":
        add_product()
    elif choice == "2":
        purchase()
    elif choice == "3":
        break
    else:
        print("Invalid choice")
```

```python
def play_game():
    score = 0
    decisions = []

    print("\nWelcome to CAMPUS SURVIVAL ADVENTURE!")
    print("Your choices will shape your future.\n")

    # Scenario 1
    print("Scenario 1: You are late for class.")
    print("A. Run to class")
    print("B. Skip class and go to cafeteria")

    choice1 = input("Choose A or B: ").upper()
    decisions.append(choice1)

    if choice1 == "A":
        score += 10
        print("You reach class on time. Good job!\n")
    else:
        score -= 5
        print("You miss an important lecture.\n")

    # Scenario 2
    print("Scenario 2: You have an assignment due.")
    print("A. Complete it yourself")
    print("B. Copy from a friend")

    choice2 = input("Choose A or B: ").upper()
    decisions.append(choice2)

    if choice2 == "A":
        score += 15
        print("Your hard work pays off!\n")
    else:
        score -= 10
        print("You get caught copying.\n")

    # Scenario 3
    print("Scenario 3: You find a lost wallet on campus.")
    print("A. Return it to office")
    print("B. Keep the money")

    choice3 = input("Choose A or B: ").upper()
    decisions.append(choice3)

    if choice3 == "A":
        score += 20
        print("You are rewarded for honesty.\n")
    else:
        score -= 20
        print("You feel guilty later.\n")

    # Ending
    print("----- GAME OVER -----")
    print("Your decisions:", decisions)
    print("Final Score:", score)

    if score >= 30:
        print("Ending: ⭐ Responsible Student ⭐")
    elif score >= 10:
        print("Ending: 🙂 Average Campus Life 🙂")
    else:
        print("Ending: ⚠ Needs Improvement ⚠")

# Replay loop
while True:
    play_game()
    replay = input("\nDo you want to play again? (yes/no): ").lower()
    if replay != "yes":
        print("Thanks for playing!")
        break
```
