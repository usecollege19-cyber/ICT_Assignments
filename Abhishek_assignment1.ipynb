# 1.Grocery Billing

```python
p1=int(input("Enter the price of item1 :"))
p2=int(input("Enter the price of item2 :"))
q1=int(input("Enter the quantity of item1 :"))
q2=int(input("Enter the quantity of item2 :"))
total=p1*q1+p2*q2
print('Total Bill Amount= ',total)
```

# 3. Salary Slip


```python
s=int(input("Enter the salary :"))
hra=0.20*s
da=0.10*s
ns=s+hra+da
print('Net Salary=',ns)
```

# 4.Simple Interest

```python
p=int(input("Enter the principal amount :"))
r=int(input("Enter the rate of interest :"))
t=int(input("Enter the time period (in years) :"))
si=(p*r*t)/100
print('Simple Interest=',si)
```

# 7.Discount Calculation

```python
p=int(input('Enter the original price: '))
d = 0.10 * p
dp = p - d
print('Discounted Price = ', dp)
```

# 8.Area of a triangle(Using Heron's Formula)

```python
a=int(input("Enter the length of side1 :"))
b=int(input("Enter the length of side2 :"))
c=int(input("Enter the length of side3 :"))
s=(a+b+c)/2
area=(s*(s-a)*(s-b)*(s-c))**0.5
print('Area of the triangle=',area)
```

# 9.Paint cost estimator

```python
l=int(input('Enter wall length : '))
w=int(input('Enter wall width : '))
win=(1.5*1.2)*2
a=(l*w)-win
pc=a*25
print('Total area to be painted = ',a,'sq.m')
print('Total painting cost : ',pc)
```
